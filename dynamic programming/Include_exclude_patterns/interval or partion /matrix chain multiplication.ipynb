```python
from sys import stdin

def mcm(p):

    # recur(i, j)
    # minimum cost to multiply matrices Ai -> Aj
    def recur(i, j):

        # only one matrix
        if i == j:
            return 0

        mini = float('inf')

        # try every partition
        for k in range(i, j):

            cost = (
                recur(i, k)                 # left part
                + recur(k + 1, j)          # right part
                + p[i - 1] * p[k] * p[j]   # merge cost
            )

            mini = min(mini, cost)

        return mini

    n = len(p)

    return recur(1, n - 1)


# number of matrices
n = int(stdin.readline().strip())

# dimensions array
p = list(map(int, stdin.readline().split()))

print(mcm(p))
```

```python
from sys import stdin

def mcm(p):

    n = len(p)

    # dp[i][j] = minimum cost to multiply Ai -> Aj
    dp = [[-1 for _ in range(n)] for _ in range(n)]

    def recur(i, j):

        # only one matrix
        if i == j:
            return 0

        # already computed
        if dp[i][j] != -1:
            return dp[i][j]

        mini = float('inf')

        # try every partition
        for k in range(i, j):

            cost = (
                recur(i, k)                 # left subchain
                + recur(k + 1, j)          # right subchain
                + p[i - 1] * p[k] * p[j]   # merge cost
            )

            mini = min(mini, cost)

        dp[i][j] = mini

        return dp[i][j]

    return recur(1, n - 1)


# number of matrices
n = int(stdin.readline().strip())

# dimensions array
p = list(map(int, stdin.readline().split()))

print(mcm(p))
```

```python
#tabulation method 
from sys import stdin

def mcm(p):

    n = len(p)

    # dp[i][j] = minimum cost to multiply Ai -> Aj
    dp = [[0 for _ in range(n)] for _ in range(n)]

    #filling diagonally upwards
    # chain length
    for length in range(2, n): #length fixes diagonal shape

        # starting index
        for i in range(1, n - length + 1): #i slides the window across matrices

            # ending index
            j = i + length - 1 #end of each diagonal 

            dp[i][j] = float('inf')

            # try every partition  #fill all the diagonals 
            for k in range(i, j):

                cost = (
                    dp[i][k]                 # left subchain
                    + dp[k + 1][j]          # right subchain
                    + p[i - 1] * p[k] * p[j]   # merge cost
                )

                dp[i][j] = min(dp[i][j], cost)

    return dp[1][n - 1]


# number of matrices
n = int(stdin.readline().strip())

# dimensions array
p = list(map(int, stdin.readline().split()))

print(mcm(p))
```
