```python
def lcs(s1, s2, i=0, j=0):
    # Base case
    if i == len(s1) or j == len(s2):
        return 0

    # Case 1: characters match
    if s1[i] == s2[j]:
        return 1 + lcs(s1, s2, i+1, j+1)

    # Case 2: mismatch → explore both
    return max(
        lcs(s1, s2, i+1, j),
        lcs(s1, s2, i, j+1)
    )
```

```python
def lcs(s1, s2):
    n, m = len(s1), len(s2)
    dp = [[-1] * m for _ in range(n)]

    def solve(i, j):
        # Base case
        if i == n or j == m:
            return 0

        # Return cached result
        if dp[i][j] != -1:
            return dp[i][j]

        # Match
        if s1[i] == s2[j]:
            dp[i][j] = 1 + solve(i+1, j+1)

        # Mismatch
        else:
            dp[i][j] = max(
                solve(i+1, j),
                solve(i, j+1)
            )

        return dp[i][j]

    return solve(0, 0)
```
