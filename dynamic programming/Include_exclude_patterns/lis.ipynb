```python
def lis(arr, index, prev, dp):
    # Base case
    if index == len(arr):
        return 0

    # Check DP
    if dp[index][prev + 1] != -1:
        return dp[index][prev + 1]

    # Not take
    not_take = lis(arr, index + 1, prev, dp)

    # Take
    take = 0
    if prev == -1 or arr[index] > arr[prev]:
        take = 1 + lis(arr, index + 1, index, dp)

    # Store result
    dp[index][prev + 1] = max(take, not_take)
    return dp[index][prev + 1]


# Driver
arr = [10, 9, 2, 5, 3, 7, 101, 18]
n = len(arr)

# DP table (n x (n+1))
dp = [[-1] * (n + 1) for _ in range(n)]

print(lis(arr, 0, -1, dp))
```
