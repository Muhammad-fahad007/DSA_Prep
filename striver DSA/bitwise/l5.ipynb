# Single Number (Bit Manipulation)

These notes explain how to solve the **Single Number** problem.

## Problem Statement

Given an integer array where every element appears **exactly twice** except for **one element**, find the element that appears only once.

**Example**

```python
nums = [4, 1, 2, 1, 2]
Output: 4
```


# Approach 1: Brute Force (Hash Map)

## Idea

Count the frequency of every element using a dictionary (hash map).

Then iterate through the dictionary and return the element whose frequency is **1**.

### Algorithm

1. Create an empty dictionary.
2. Count the occurrence of every number.
3. Traverse the dictionary.
4. Return the key whose frequency is 1.

### Complexity

- **Time:** \(O(N)\) on average (or \(O(N\log M)\) depending on map implementation)
- **Space:** \(O(M)\), where \(M\) is the number of unique elements.


````
def singleNumber_hash(nums):
    freq = {}

    for num in nums:
        freq[num] = freq.get(num, 0) + 1

    for num, count in freq.items():
        if count == 1:
            return num


nums = [4, 1, 2, 1, 2]
print(singleNumber_hash(nums))
````

# Approach 2: Optimal (Bit Manipulation using XOR)

## XOR Properties

```text
a ^ a = 0
a ^ 0 = a
```

Since every duplicate appears exactly twice,

```text
a ^ a = 0
```

All duplicate numbers cancel each other.

Only the unique number remains.

### Algorithm

1. Initialize `ans = 0`.
2. XOR every element with `ans`.
3. Return `ans`.

This works because duplicate values become **0**, leaving only the single element.


````
def singleNumber(nums):
    ans = 0

    for num in nums:
        ans ^= num

    return ans


nums = [4, 1, 2, 1, 2]
print(singleNumber(nums))
````

Notice how:

```text
1 ^ 1 = 0

```

Only **4** remains.


# Complexity Analysis

## Brute Force

- **Time:** \(O(N)\) average
- **Space:** \(O(M)\)

## Optimal XOR Solution

- **Time:** \(O(N)\)
- **Space:** \(O(1)\)

---

# Key Takeaway

The XOR approach is the most efficient because:

- Duplicate elements cancel each other.
- No extra data structures are required.
- Only one pass through the array is needed.
- Constant extra space.
