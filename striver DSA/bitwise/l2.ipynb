# Bit Manipulation Tricks (Interview Notes)
Based on the requested topics with Python examples.

## Swap Two Numbers (XOR)

```
a=5
b=9
print("Before:",a,b)
a^=b
b^=a
a^=b
print("After :",a,b)

# Preferred in Python
x,y=5,9
x,y=y,x
print("Pythonic:",x,y)
```

## Check if i-th Bit is Set

```
n=13   #1101
i=2
print(bool(n & (1<<i)))
print(((n>>i)&1)==1)
```

## Set the i-th Bit

```
n=10   #1010
i=0
print(bin(n))
print(bin(n | (1<<i)))
```

## Clear the i-th Bit

```
n=11   #1011
i=1
print(bin(n))
print(bin(n & ~(1<<i)))
```

## Toggle the i-th Bit

```
n=10   #1010
i=0
print(bin(n))
print(bin(n ^ (1<<i)))
```

## Remove Last Set Bit

```
n=12
print(bin(n))
print(bin(n & (n-1)))
```

## Power of Two

```
def is_power_of_two(n):
    return n>0 and (n & (n-1))==0

for x in [1,2,4,8,10,16,18]:
    print(x,is_power_of_two(x))
```

## Count Set Bits (Brute Force)

```
def count_bits_brute(n):
    cnt=0
    while n:
        cnt += n & 1
        n >>= 1
    return cnt

print(count_bits_brute(13))
```

## Count Set Bits (Brian Kernighan)

```
def count_bits(n):
    cnt=0
    while n:
        n &= (n-1)
        cnt += 1
    return cnt

print(count_bits(13))
```

## Python Built-in

```
print((13).bit_count())
```

## Cheat Sheet


| Trick | Formula |
|---|---|
| Swap | `a ^= b; b ^= a; a ^= b` |
| Check bit | `(n >> i) & 1` |
| Set bit | `n | (1 << i)` |
| Clear bit | `n & ~(1 << i)` |
| Toggle bit | `n ^ (1 << i)` |
| Remove last set bit | `n & (n - 1)` |
| Power of Two | `n > 0 and (n & (n-1)) == 0` |
| Count set bits | `while n: n &= (n-1)` |
| Built-in | `n.bit_count()` |
