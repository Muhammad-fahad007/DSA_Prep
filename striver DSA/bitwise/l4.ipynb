# Power Set using Bit Manipulation

These notes explain how to generate the **Power Set** (all possible subsets of an array) using **bit manipulation**.


## Core Concepts

### 1. Number of Subsets
For an array of size **n**, the total number of subsets is:

\[
2^n
\]

This can be calculated efficiently using:

```python
1 << n
```

because left shifting `1` by `n` positions is equivalent to \(2^n\).

---

### 2. Bit Representation

Every integer from **0** to **2ⁿ − 1** represents one unique subset.

The **j-th bit** determines whether the element at index `j` is included:

- **1** → Include the element.
- **0** → Skip the element.

Example:

```text
nums = [1, 2, 3]

Binary   Subset
000      []
001      [1]
010      [2]
011      [1,2]
100      [3]
101      [1,3]
110      [2,3]
111      [1,2,3]
```


## Algorithm

1. Compute the total number of subsets:

```python
total = 1 << n
```

2. Iterate `i` from `0` to `total - 1`.

3. For every bit position `j` from `0` to `n-1`:

```python
if i & (1 << j):
```

4. If the condition is true, include `nums[j]` in the current subset.

5. Add the completed subset to the answer list.


````
def subsets(nums):
    n = len(nums)
    ans = []

    for i in range(1 << n):
        subset = []

        for j in range(n):
            if i & (1 << j):
                subset.append(nums[j])

        ans.append(subset)

    return ans


nums = [1, 2, 3]
print(subsets(nums))
````

## Dry Run

For:

```python
nums = [1, 2, 3]
```

### i = 5

Binary representation:

```text
5 = 101
```

Check each bit:

| Bit Position | Bit | Element | Include? |
|--------------|-----|---------|----------|
| 0 | 1 | 1 | ✅ |
| 1 | 0 | 2 | ❌ |
| 2 | 1 | 3 | ✅ |

Resulting subset:

```python
[1, 3]
```


## Complexity Analysis

### Time Complexity

There are **2ⁿ** subsets, and constructing each subset requires checking **n** bits.

\[
O(n \cdot 2^n)
\]

---

### Space Complexity

Storing every subset requires approximately:

\[
O(n \cdot 2^n)
\]

since there are **2ⁿ** subsets and each subset may contain up to **n** elements.
