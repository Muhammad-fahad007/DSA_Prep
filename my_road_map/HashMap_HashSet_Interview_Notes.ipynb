## Contains Duplicate — 217 — Easy

**LeetCode:** [Contains Duplicate](https://leetcode.com/problems/contains-duplicate/)

### Recognition
Need to know whether any value has appeared before.

### Intuition
Maintain a set of values already seen. If the current value is already in the set, a duplicate exists.

### Code
```python
def containsDuplicate(nums):
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False
```

### Complexity
Time: **O(n)** average. Space: **O(n)**.

### Tradeoffs / Interview Notes
Set gives O(1) average membership and is exactly what is needed; no reason to store indices or frequencies.


## Valid Anagram — 242 — Easy

**LeetCode:** [Valid Anagram](https://leetcode.com/problems/valid-anagram/)

### Recognition
Need to compare character frequencies.

### Intuition
Anagrams have identical counts for every character. Store character → frequency and compare.

### Code
```python
def isAnagram(s, t):
    if len(s) != len(t):
        return False

    freq = {}
    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in t:
        freq[ch] = freq.get(ch, 0) - 1
        if freq[ch] < 0:
            return False

    return all(v == 0 for v in freq.values())
```

### Complexity
Time: **O(n)**. Space: **O(k)** where k is the character set.

### Tradeoffs / Interview Notes
If alphabet is fixed, an array of counts can be simpler/faster than a dict. Avoid comparing `dict.values()` because character-to-count correspondence matters.


## Two Sum — 1 — Easy

**LeetCode:** [Two Sum](https://leetcode.com/problems/two-sum/)

### Recognition
At each element, need to know whether its complement has appeared earlier.

### Intuition
Traverse left-to-right. For current `x`, required value is `target - x`. Store previous value → index so lookup is O(1).

### Code
```python
def twoSum(nums, target):
    seen = {}
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            return [seen[need], i]
        seen[x] = i
    return []
```

### Complexity
Time: **O(n)** average. Space: **O(n)**.

### Tradeoffs / Interview Notes
One-pass HashMap is preferable to sorting because sorting loses original indices and costs O(n log n).


## Ransom Note — 383 — Easy

**LeetCode:** [Ransom Note](https://leetcode.com/problems/ransom-note/)

### Recognition
Need enough copies of each character from one string to construct another.

### Intuition
Count magazine characters, then consume counts while traversing ransom note.

### Code
```python
def canConstruct(ransomNote, magazine):
    freq = {}
    for ch in magazine:
        freq[ch] = freq.get(ch, 0) + 1

    for ch in ransomNote:
        if freq.get(ch, 0) == 0:
            return False
        freq[ch] -= 1

    return True
```

### Complexity
Time: **O(m+n)**. Space: **O(k)**.

### Tradeoffs / Interview Notes
Frequency subtraction directly models resource consumption. With a fixed alphabet, a count array is an alternative.


## Majority Element — 169 — Easy

**LeetCode:** [Majority Element](https://leetcode.com/problems/majority-element/)

### Recognition
Need an element occurring more than n/2 times.

### Intuition
HashMap frequency works directly: count each value and return the one whose count exceeds n//2.

### Code
```python
def majorityElement(nums):
    freq = {}
    for x in nums:
        freq[x] = freq.get(x, 0) + 1
        if freq[x] > len(nums) // 2:
            return x
```

### Complexity
HashMap approach: Time **O(n)**, Space **O(n)**.

### Tradeoffs / Interview Notes
Because the problem guarantees a majority, **Boyer-Moore Voting** achieves O(n) time and O(1) space. Know the tradeoff even if HashMap is easier to derive.


## Group Anagrams — 49 — Medium

**LeetCode:** [Group Anagrams](https://leetcode.com/problems/group-anagrams/)

### Recognition
Need to group strings with the same character composition.

### Intuition
Create a canonical signature for each word; use signature → list of words.

### Code
```python
def groupAnagrams(strs):
    groups = {}
    for word in strs:
        count = [0] * 26
        for ch in word:
            count[ord(ch) - ord('a')] += 1
        key = tuple(count)

        if key not in groups:
            groups[key] = []
        groups[key].append(word)

    return list(groups.values())
```

### Complexity
Time: **O(n·k)** for n strings of average length k. Space: **O(n·k)** including output/group storage.

### Tradeoffs / Interview Notes
Sorted-word key is simpler but costs O(k log k) per word. Frequency signature is O(k) when alphabet is fixed.


## Top K Frequent Elements — 347 — Medium

**LeetCode:** [Top K Frequent Elements](https://leetcode.com/problems/top-k-frequent-elements/)

### Recognition
Need elements with highest frequencies; frequencies are bounded by n.

### Intuition
Count frequencies, then use frequency as a bucket index. Iterate buckets from high to low until k elements are collected.

### Code
```python
def topKFrequent(nums, k):
    freq = {}
    for x in nums:
        freq[x] = freq.get(x, 0) + 1

    bucket = [[] for _ in range(len(nums) + 1)]
    for x, count in freq.items():
        bucket[count].append(x)

    ans = []
    for f in range(len(bucket) - 1, 0, -1):
        ans.extend(bucket[f])
        if len(ans) >= k:
            return ans[:k]
    return ans
```

### Complexity
Time: **O(n)** average. Space: **O(n)**.

### Tradeoffs / Interview Notes
Bucket sort exploits bounded frequency `1..n`. A min-heap gives O(n log k) and is preferable when k is small or frequencies are not naturally bounded.


## Longest Consecutive Sequence — 128 — Medium

**LeetCode:** [Longest Consecutive Sequence](https://leetcode.com/problems/longest-consecutive-sequence/)

### Recognition
Need the longest consecutive integer sequence; membership checks should be O(1).

### Intuition
Put all numbers in a set. Only start a sequence when `n-1` is absent. Then repeatedly check `n+1`.

### Code
```python
def longestConsecutive(nums):
    numset = set(nums)
    longest = 0

    for n in numset:
        if n - 1 not in numset:
            length = 1
            while n + 1 in numset:
                n += 1
                length += 1
            longest = max(longest, length)

    return longest
```

### Complexity
Time: **O(n)** average. Space: **O(n)**.

### Tradeoffs / Interview Notes
The `n-1 not in set` condition prevents repeatedly scanning the middle of the same sequence. Without it, the algorithm can degrade to O(n²).


## Contains Duplicate II — 219 — Easy

**LeetCode:** [Contains Duplicate II](https://leetcode.com/problems/contains-duplicate-ii/)

### Recognition
Need a duplicate and its previous index to check whether distance ≤ k.

### Intuition
Store value → latest index. On seeing a duplicate, compare current index with the latest stored index, then update to current index.

### Code
```python
def containsNearbyDuplicate(nums, k):
    last = {}
    for i, x in enumerate(nums):
        if x in last and i - last[x] <= k:
            return True
        last[x] = i
    return False
```

### Complexity
Time: **O(n)** average. Space: **O(n)**.

### Tradeoffs / Interview Notes
Latest index is enough because for a fixed current index, the latest previous occurrence gives the smallest index distance.


## Isomorphic Strings — 205 — Easy

**LeetCode:** [Isomorphic Strings](https://leetcode.com/problems/isomorphic-strings/)

### Recognition
Need a one-to-one character mapping in both directions.

### Intuition
Store `s → t` and `t → s`. Both maps prevent two source characters mapping to one target or vice versa.

### Code
```python
def isIsomorphic(s, t):
    s_to_t = {}
    t_to_s = {}

    for a, b in zip(s, t):
        if a in s_to_t and s_to_t[a] != b:
            return False
        if b in t_to_s and t_to_s[b] != a:
            return False

        s_to_t[a] = b
        t_to_s[b] = a

    return True
```

### Complexity
Time: **O(n)**. Space: **O(k)**.

### Tradeoffs / Interview Notes
Two maps make the bijection explicit. A single map is insufficient because `a→x` and `b→x` would otherwise incorrectly pass.


## Subarray Sum Equals K — 560 — Medium

**LeetCode:** [Subarray Sum Equals K](https://leetcode.com/problems/subarray-sum-equals-k/)

### Recognition
Count subarrays with sum exactly k, including negative values.

### Intuition
Use prefix sum `S`. If an earlier prefix sum was `S-k`, the elements between it and the current position sum to k. Store prefix_sum → frequency.

### Code
```python
def subarraySum(nums, k):
    prefix_freq = {0: 1}
    summ = 0
    ans = 0

    for x in nums:
        summ += x
        ans += prefix_freq.get(summ - k, 0)
        prefix_freq[summ] = prefix_freq.get(summ, 0) + 1

    return ans
```

### Complexity
Time: **O(n)** average. Space: **O(n)**.

### Tradeoffs / Interview Notes
`{0: 1}` represents the empty prefix before the array, allowing subarrays starting at index 0. Store frequency, not latest index, because multiple previous prefixes can produce different valid subarrays.


## Happy Number — 202 — Easy

**LeetCode:** [Happy Number](https://leetcode.com/problems/happy-number/)

### Recognition
A process repeatedly transforms a state and may enter a cycle.

### Intuition
Store each generated number in a set. If 1 appears, succeed. If a number repeats, a cycle exists and the number is not happy.

### Code
```python
def isHappy(n):
    seen = set()

    while n != 1:
        if n in seen:
            return False
        seen.add(n)

        total = 0
        while n:
            digit = n % 10
            total += digit * digit
            n //= 10

        n = total

    return True
```

### Complexity
Time: effectively **O(log n)** per transformation with a small bounded state cycle; Space: **O(number of states)**.

### Tradeoffs / Interview Notes
HashSet cycle detection is simple. Floyd's tortoise-and-hare can detect cycles with O(1) extra space, but is less direct here.


## Valid Sudoku — 36 — Medium

**LeetCode:** [Valid Sudoku](https://leetcode.com/problems/valid-sudoku/)

### Recognition
Need uniqueness across rows, columns, and 3×3 boxes.

### Intuition
Maintain three collections: row → seen values, column → seen values, box → seen values. A box ID can be `(i//3, j//3)` or a flattened integer.

### Code
```python
def isValidSudoku(board):
    rows = [set() for _ in range(9)]
    cols = [set() for _ in range(9)]
    boxes = [set() for _ in range(9)]

    for i in range(9):
        for j in range(9):
            x = board[i][j]
            if x == '.':
                continue

            box = (i // 3) * 3 + (j // 3)

            if x in rows[i] or x in cols[j] or x in boxes[box]:
                return False

            rows[i].add(x)
            cols[j].add(x)
            boxes[box].add(x)

    return True
```

### Complexity
Fixed 9×9 board: **O(1)** time/space. General n×n board: **O(n²)** time/space.

### Tradeoffs / Interview Notes
Because dimensions are fixed, lists of sets are cleaner than dictionaries/defaultdicts. The calculated box ID is a reusable grouping technique.


## 4Sum II — 454 — Medium

**LeetCode:** [4Sum II](https://leetcode.com/problems/4sum-ii/)

### Recognition
Four separate arrays satisfy `A+B+C+D=0`; brute force is O(n⁴). Look for a way to split the equation.

### Intuition
Group into `(A+B) + (C+D)=0`. Precompute all A+B sums as `sum → frequency`; for each C+D, add the frequency of `-(C+D)`.

### Code
```python
def fourSumCount(A, B, C, D):
    freq = {}

    for a in A:
        for b in B:
            s = a + b
            freq[s] = freq.get(s, 0) + 1

    ans = 0
    for c in C:
        for d in D:
            ans += freq.get(-(c + d), 0)

    return ans
```

### Complexity
Time: **O(n²)**. Space: **O(n²)**.

### Tradeoffs / Interview Notes
3 loops + lookup is O(n³), but pair-sum precomputation removes another loop. This is a classic 'precompute once, reuse many times' optimization. Think of it as Two Sum on pair sums.


## Longest Substring Without Repeating Characters — 3 — Medium

**LeetCode:** [Longest Substring Without Repeating Characters](https://leetcode.com/problems/longest-substring-without-repeating-characters/)

### Recognition
Need the longest contiguous window satisfying a property: all characters are unique.

### Intuition
Expand `right`. Store character → latest index. If a duplicate is inside the current window, jump `left` to one past its latest index. Use `max` window length.

### Code
```python
def lengthOfLongestSubstring(s):
    last_seen = {}
    left = 0
    ans = 0

    for right, ch in enumerate(s):
        if ch in last_seen:
            left = max(left, last_seen[ch] + 1)

        last_seen[ch] = right
        ans = max(ans, right - left + 1)

    return ans
```

### Complexity
Time: **O(n)**. Space: **O(min(n, charset))**.

### Tradeoffs / Interview Notes
The `max` prevents `left` from moving backward when an old duplicate is outside the current window. HashSet can detect duplicates, but HashMap with latest index lets us jump directly.


## 4Sum — 18 — Medium

**LeetCode:** [4Sum](https://leetcode.com/problems/4sum/)

### Recognition
One sorted array; find unique quadruplets summing to target. Think fixed elements + ordered remaining search.

### Intuition
Sort. Fix `i` and `j`; use `left/right` pointers for the remaining two. If sum is too small, move left; if too large, move right. Skip duplicates at every level.

### Code
```python
def fourSum(nums, target):
    nums.sort()
    n = len(nums)
    ans = []

    for i in range(n - 3):
        if i > 0 and nums[i] == nums[i - 1]:
            continue

        for j in range(i + 1, n - 2):
            if j > i + 1 and nums[j] == nums[j - 1]:
                continue

            left, right = j + 1, n - 1

            while left < right:
                total = nums[i] + nums[j] + nums[left] + nums[right]

                if total < target:
                    left += 1
                elif total > target:
                    right -= 1
                else:
                    ans.append([nums[i], nums[j], nums[left], nums[right]])
                    left += 1
                    right -= 1

                    while left < right and nums[left] == nums[left - 1]:
                        left += 1
                    while left < right and nums[right] == nums[right + 1]:
                        right -= 1

    return ans
```

### Complexity
Time: **O(n³)** after sorting (sorting is O(n log n)). Extra space: **O(1)** excluding output.

### Tradeoffs / Interview Notes
Sorting enables monotonic two-pointer movement. Pair-sum HashMap approaches exist, but can require O(n²) memory and careful index/duplicate handling. For interview purposes, sorted + two pointers is the standard robust approach.


# Google L4 Interview Recognition Checklist

Before coding any unfamiliar problem:

### 1. What is the brute force?
Can I describe the obvious solution and its complexity?

### 2. What work is repeated?
Look for:
- repeated membership checks
- repeated frequency calculations
- repeated pair/sum calculations
- repeated scans of a window

### 3. What should I remember?
- **Existence** → Set
- **Count** → Map frequency
- **Previous position** → Map index
- **Complement** → Map lookup
- **State history** → Set
- **Prefix history** → Map frequency
- **Pair results reused later** → precompute pair sums

### 4. What invariant am I maintaining?
Examples:
- Sliding window contains only unique characters.
- Current prefix sum history is stored with frequencies.
- `left/right` enclose all remaining candidates for sorted 4Sum.

### 5. Can I explain the tradeoff?
Google interviewers care about *why* the optimization works, not just the final code.

## High-value patterns from this set

### HashMap
`value → frequency`, `value → index`, `key → group`, `prefix_sum → frequency`

### HashSet
existence, duplicate detection, sequence starts, cycle detection

### Precomputation
Compute reusable information once, store it, and avoid repeating work.

### Sliding Window
Maintain a contiguous window and repair it when its invariant breaks.

### Two Pointers
Sorting gives an ordered search space; move pointers based on whether the current sum is too small/large.
