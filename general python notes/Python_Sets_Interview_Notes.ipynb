# Python Sets - Interview Notes
Comprehensive notes for LeetCode and coding interviews.

## 1. What is a Set?
- Unordered collection of **unique** elements.
- Implemented as a **hash table**.
- Average-case `O(1)` lookup, insertion, and deletion.

## 2. Creating Sets

````
# Empty set
s = set()

# Non-empty set
s = {1, 2, 3}

# {} creates a dictionary, not a set
d = {}
print(type(s), type(d))
````

## 3. Adding Elements

````
s = {1, 2}
s.add(3)
print(s)
````

## 4. Removing Elements

````
s = {1, 2, 3}

s.remove(2)      # Raises KeyError if absent
s.discard(10)    # Safe if absent

x = s.pop()      # Removes an arbitrary element
print("Popped:", x)

s.clear()        # Removes all elements
print(s)
````

## 5. Membership Testing

````
s = {1, 2, 3}

print(2 in s)
print(5 not in s)
````

## 6. Length

````
s = {1,2,3}
print(len(s))
````

## 7. Iteration

````
for x in {10,20,30}:
    print(x)
````

## 8. Convert List ↔ Set

````
nums = [1,2,2,3,3]

print(set(nums))
print(list(set(nums)))
````

## 9. Set Operations

````
a = {1,2,3}
b = {3,4,5}

print("Union:", a | b)
print("Intersection:", a & b)
print("Difference:", a - b)
print("Symmetric Difference:", a ^ b)
````

## 10. Subset / Superset / Disjoint

````
a = {1,2}
b = {1,2,3}

print(a <= b)
print(a < b)

print(b >= a)
print(b > a)

print(a.isdisjoint({5,6}))
````

## 11. Additional Useful Methods

````
a = {1,2}
a.update([3,4])
print(a)

b = {2,3,4}

print(a.intersection(b))
print(a.difference(b))
print(a.symmetric_difference(b))
````

## 12. Interview Patterns

````
# Remove duplicates
nums = [1,2,2,3]
print(list(set(nums)))

# Contains Duplicate
nums = [1,2,3,1]
print(len(nums) != len(set(nums)))

# Seen before
seen = set()
for x in nums:
    if x in seen:
        print("Duplicate:", x)
        break
    seen.add(x)

# Unique characters
s = "leetcode"
print(len(s) == len(set(s)))
````

## 13. Time Complexity

| Operation | Complexity |
|---|---|
| `add()` | O(1) average |
| `remove()` | O(1) average |
| `discard()` | O(1) average |
| `in` / `not in` | O(1) average |
| `pop()` | O(1) average |
| `len()` | O(1) |
| Union (`|`) | O(len(a)+len(b)) |
| Intersection (`&`) | O(min(len(a), len(b))) |
| Difference (`-`) | O(len(a)) |

## 14. Common Pitfalls

````
# {} is a dictionary
print(type({}))
print(type(set()))

# Sets remove duplicates
print({1,1,2,2,3})

# Sets are unordered
s = {1,2,3}
# s[0]   # TypeError

# Only hashable objects
valid = {(1,2)}
print(valid)

# invalid = {[1,2]}  # TypeError
````
