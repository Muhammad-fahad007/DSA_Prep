# Python Bit Manipulation Notes (Interview + LeetCode)
Comprehensive notes covering the most important bit manipulation concepts, tricks, and interview patterns.

## 1. Binary Representation

````python
n = 13

print(bin(n))          # 0b1101
print(format(n, "b"))  # 1101
````

## 2. Bitwise Operators

````python
a = 5   # 0101
b = 3   # 0011

print(a & b)   # AND
print(a | b)   # OR
print(a ^ b)   # XOR
print(~a)      # NOT
print(a << 1)  # Left shift
print(a >> 1)  # Right shift
````

## 3. Left & Right Shift

````python
n = 5
print(n << 2)   # multiply by 4

n = 20
print(n >> 2)   # divide by 4 (non-negative integers)
````

## 4. Check Odd / Even

````python
n = 7
print(n & 1)    # 1 => odd, 0 => even
````

## 5. Set, Clear, Toggle, Check ith Bit

````python
n = 10
i = 1

print(n | (1 << i))       # Set bit
print(n & ~(1 << i))      # Clear bit
print(n ^ (1 << i))       # Toggle bit
print((n >> i) & 1)       # Check bit
````

## 6. Remove Lowest Set Bit

````python
n = 12
print(bin(n))
print(bin(n & (n-1)))
````

## 7. Isolate Lowest Set Bit

````python
n = 12
print(bin(n & -n))
````

## 8. Count Set Bits

````python
def count_bits(n):
    cnt = 0
    while n:
        n &= (n-1)
        cnt += 1
    return cnt

print(count_bits(13))
print((13).bit_count())
````

## 9. Power of Two

````python
def is_power_of_two(n):
    return n > 0 and (n & (n-1)) == 0

print(is_power_of_two(16))
print(is_power_of_two(18))
````

## 10. Bit Length

````python
print((13).bit_length())
````

## 11. XOR Properties

````python
# a ^ a = 0
# a ^ 0 = a
# a ^ b ^ a = b

nums = [2,3,2]
ans = 0
for x in nums:
    ans ^= x
print(ans)
````

## 12. XOR Swap

````python
a, b = 5, 9
a ^= b
b ^= a
a ^= b
print(a, b)

# Preferred:
a, b = b, a
````

## 13. Common Interview Patterns

````python
# Single Number
ans = 0
for x in [2,2,1]:
    ans ^= x
print(ans)

# Missing Number
def missing(nums):
    ans = len(nums)
    for i, v in enumerate(nums):
        ans ^= i ^ v
    return ans

# Counting Bits
def countBits(n):
    dp = [0]*(n+1)
    for i in range(1, n+1):
        dp[i] = dp[i>>1] + (i & 1)
    return dp
````

## Time Complexity

| Operation | Complexity |
|---|---|
| `&`, `|`, `^`, `~` | O(1) |
| `<<`, `>>` | O(1) |
| `bit_count()` | O(number of machine words) |
| Brian Kernighan | O(number of set bits) |
