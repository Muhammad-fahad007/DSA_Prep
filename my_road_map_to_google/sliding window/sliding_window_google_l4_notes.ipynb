# 1. Core Sliding Window Mental Model

A window is a contiguous portion:

```text
[ i ........ j ]
```

Move `j` right to expand. Move `i` right to shrink.

## Fixed-size window

The window size is exactly `k`.

```python
for j in range(len(nums)):
    add(nums[j])

    if j >= k:
        remove(nums[j-k])

    if j >= k-1:
        evaluate()
```

## Variable-size window

```python
for j in range(len(nums)):
    add(nums[j])

    while window_is_invalid:
        remove(nums[i])
        i += 1

    evaluate()
```

## Recognition question

Don't ask only:

> Is this contiguous?

Ask:

> **If the window becomes invalid, can I reliably repair it by moving `left` forward?**

For positive numbers with sum/product constraints, the answer is often yes because the quantity is monotonic.

With arbitrary negative numbers, exact-sum problems often point toward **Prefix Sum + HashMap** instead.

# 2. 1456 — Maximum Number of Vowels in a Substring of Given Length

**Pattern:** Fixed-size Sliding Window

The window must always contain exactly `k` characters.

Instead of recounting vowels for every substring:

- add the new right character
- remove the old left character
- maintain the current vowel count

**Complexity:** `O(n)` time, `O(1)` space.

### Recognition

> Maximum/minimum over every contiguous substring of exactly `k` elements

→ **Fixed Sliding Window**.

# 3. 209 — Minimum Size Subarray Sum

**Pattern:** Variable Sliding Window + positive sum

Find the minimum-length contiguous subarray whose sum is at least `target`.

Because all numbers are positive:

- expanding right increases the sum
- moving left decreases the sum

When:

```text
window_sum >= target
```

the window is valid, so shrink aggressively:

```python
while window_sum >= target:
    ans = min(ans, j-i+1)
    window_sum -= nums[i]
    i += 1
```

### Key rule

**Minimum length:** expand until valid → shrink as much as possible.

**Complexity:** `O(n)` time, `O(1)` space.

# 4. 1004 — Max Consecutive Ones III

**Pattern:** Variable Window + constraint

We can flip at most `k` zeroes.

Window condition:

```text
zero_count <= k
```

When `zero_count > k`, shrink from the left until valid.

Then:

```python
ans = max(ans, j-i+1)
```

### Recognition

> Longest contiguous range with an **at most K** constraint

→ Variable Sliding Window.

# 7. 424 — Longest Repeating Character Replacement

**Pattern:** Variable Window + frequency map

For a window of length `L`, let `max_freq` be the frequency of its most common character.

Characters that need replacement:

```text
L - max_freq
```

The window is valid when:

```text
L - max_freq <= k
```

or:

```python
(j-i+1) - max_freq <= k
```

### Key intuition

We don't care which character to replace. We keep the most frequent character and replace everything else.

**Complexity:** `O(n)` time, `O(Alphabet)` space.

# 9. 713 — Subarray Product Less Than K

**Pattern:** Variable Window + product + counting

Maintain product.

Valid condition:

```text
product < k
```

When:

```text
product >= k
```

shrink from the left.

### The important counting trick

If `[i ... j]` is valid, then the number of valid subarrays ending at `j` is:

```text
j - i + 1
```

They are:

```text
[i...j]
[i+1...j]
...
[j...j]
```

Therefore:

```python
ans += j-i+1
```

### Critical edge case

If:

```python
k <= 1:
    return 0
```

because all positive products are at least `1`.

### Why positivity matters

Negative numbers destroy the monotonic product behavior needed by this window technique.

# 10. 1423 — Maximum Points You Can Obtain from Cards

**Pattern:** Complementary Sliding Window

You take exactly `k` cards from the ends.

Greedy idea:

> Take the larger end card.

This can fail.

Instead, notice:

```text
cards taken = k
cards left = n-k
```

The cards left behind form **one contiguous middle subarray**.

Therefore:

```text
maximum taken sum
=
total sum
-
minimum contiguous sum of length n-k
```

So the problem becomes a **fixed-size Sliding Window** problem.

### Recognition lesson

When a problem says:

> select from the ends

ask:

> **Can I solve the complement — the contiguous middle that remains?**

# 11. 1658 — Minimum Operations to Reduce X to Zero

**Pattern:** Transform → longest Sliding Window

We remove elements from the left/right until their sum is `x`.

Let:

```text
total = sum(nums)
```

If removed sum is `x`, the elements kept in the middle sum to:

```text
total - x
```

So instead find:

> **The longest contiguous subarray with sum `total-x`.**

Then:

```text
answer = n - longest_length
```

### Important edge cases

If:

```text
total-x < 0
```

→ impossible → `-1`

If:

```text
total-x == 0
```

→ keep the empty subarray → answer `n`.

### Transformation

```text
remove from ends
       ↓
keep middle
       ↓
longest valid contiguous subarray

# 12. 1248 — Count Number of Nice Subarrays

**Pattern:** Sliding Window + counting

A nice subarray contains exactly `k` odd numbers.

Maintain:

```text
odd_count
```

Expand right.

If:

```text
odd_count > k
```

shrink from the left.

The tricky part is counting **all** valid starting positions.

Once the window has exactly `k` odds, there may be several leading even numbers that can be removed while keeping exactly `k` odds.

Another standard way to express the counting idea is:

```text
exactly(k) = atMost(k) - atMost(k-1)
```

### Recognition

When asked to count subarrays with an exact condition, consider whether:

```text
exactly K
=
at most K
-
at most K-1
```


# 🧠 Sliding Window Recognition Cheat Sheet

## Fixed window

Look for:

- exactly `k` elements
- maximum/minimum over every window of size `k`
- substring of length `k`

Skeleton:

```python
for j in range(n):
    add(nums[j])

    if j >= k:
        remove(nums[j-k])

    if j >= k-1:
        evaluate()
```

## Variable window

Look for:

- longest/shortest contiguous range
- at most `K`
- budget/cost constraint
- positive-number sum/product constraint
- frequency/distinct-element constraint

Skeleton:

```python
for j in range(n):
    add(nums[j])

    while invalid:
        remove(nums[i])
        i += 1

    evaluate()
```

## Maximum length

```python
ans = max(ans, j-i+1)
```

Think:

> **Expand → repair invalidity → maximize**

## Minimum length

```python
ans = min(ans, j-i+1)
```

Think:

> **Expand → become valid → shrink aggressively**

## Counting

Ask:

> **How many valid subarrays end at `j`?**

Classic example:

```text
j-i+1
```

And for exact constraints:

```text
exactly(k) = atMost(k) - atMost(k-1)

# 🚨 When NOT to use Sliding Window

### Contiguous ≠ automatically Sliding Window

For positive numbers:

```text
window sum too large
→ remove from left
→ sum decreases predictably
```

This monotonicity makes Sliding Window possible.

With positive + negative numbers:

```text
[3, -2, 5, -1]
```

removing the left element can change the sum unpredictably.

For exact-sum problems, consider:

> **Prefix Sum + HashMap**

### Your recognition rule

> **Sliding Window needs a predictable way to repair an invalid window by moving `left` forward.**

If that property is absent, stop and consider another pattern.

# 📝 Mistakes & Lessons We Covered

1. **Shrink using the left pointer**
   ```python
   window_sum -= nums[i]
   ```
   not `nums[j]`.

2. **Use `while` when repeated shrinking may be necessary.**

3. **Check the monotonicity assumption.**
   Positive numbers are important for many sum/product windows.

4. **Don't confuse finding a valid window with counting all valid windows.**

5. **Don't default to greedy.**
   LC 1423 is a good example where a transformation is better.

6. **Don't assume every contiguous-subarray problem is Sliding Window.**

7. **For LC 713, remember the counting formula:**
   ```text
   number of valid subarrays ending at j = j-i+1
   ```

8. **For minimum-window problems, once valid, shrink aggressively.**
